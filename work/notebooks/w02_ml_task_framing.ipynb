## 2. Target or proxy

## 3. Success metric

*To evaluate clustering quality, I'll rely on the silhouette score, a numerical indicator that's higher when pages align more closely with others in their own cluster than with pages elsewhere. Beyond that metric, I'll also apply a manual review by examining each cluster's typical characteristics to check whether the groupings actually make practical sense. So success here means more than a strong numerical score. It also requires clusters that are sufficiently distinct and meaningful enough to inform real content decisions.*

## 5. Why ML beats a fixed rule here

*A fixed rule could identify a simple group, such as pages with high impressions and low CTR, but it would be difficult to write one rule that captures the many different combinations of content age, freshness, impressions, CTR, average position, engagement, and word count. Clustering can examine these signals together and help reveal recurring groups of pages that a single if-statement might miss. The goal is not to claim that ML is always better than rules, but to test whether data-driven clusters provide a more useful view of the content inventory for decision support.*
